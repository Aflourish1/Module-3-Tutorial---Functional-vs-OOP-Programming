Sort an Array of 0s, 1s, and 2s

```python
class Solution:
    def sort012(self, arr):
        
        #Sorting an array containing only 0s, 1s, and 2s in ascending order

        # Initialize counters for each number
        count0 = 0
        count1 = 0
        count2 = 0

        # Count how many times each number appears
        for num in arr:
            if num == 0:
                count0 += 1
            elif num == 1:
                count1 += 1
            elif num == 2:
                count2 += 1

        # Arrange the array in ascending order
        arr[:] = ([0] * count0) + ([1] * count1) + ([2] * count2)      

# Input to test the function
arr = [0, 1, 2, 0, 1, 2]
```

Binary Search

```python
class Solution:
    def firstSearch(self, arr, k):
        #Finds the first occurrence of k in a sorted array using binary search.
        #Returns -1 if k is not found.
         
        # Initialize the starting and ending positions
        left = 0
        right = len(arr) - 1

        # Store the first occurrence of k
        result = -1

        # Continue searching while the range is valid
        while left <= right:

            # Find the middle index
            mid = (left + right) // 2

            # Check if the middle element equals k
            if arr[mid] == k:
                result = mid

                # Continue searching the left side
                # to find an earlier occurrence
                right = mid - 1

            # If the middle element is smaller than k
            elif arr[mid] < k:
                left = mid + 1

            # If the middle element is greater than k
            else:
                right = mid - 1

        return result
```
